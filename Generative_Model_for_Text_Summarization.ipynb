```python
from transformers import pipeline

# List of models
models = [
    "facebook/bart-large-cnn", "facebook/bart-large-xsum", "google/pegasus-xsum",
    "google/pegasus-cnn_dailymail", "google/t5-small", "google/t5-base",
    "google/t5-large", "allenai/led-base-16384", "allenai/led-large-16384-arxiv",
    "bigbird-pegasus-large-arxiv", "microsoft/prophetnet-large-uncased",
    "microsoft/prophetnet-large-cnn-dailymail", "mrm8488/t5-small-finetuned-summarize-news",
    "Salesforce/bart-large-xsum-samsum", "facebook/mbart-large-50", "Yale-LILY/brio-xsum-cased",
    "pszemraj/led-large-book-summary", "facebook/bart-base", "lucadiliello/mt5-base-it-sum",
    "allenai/longformer-large-4096", "flax-community/t5-base-cnn-dm",
    "allenai/led-large-16384-arxiv", "allenai/longformer-large-4096-booksum",
    "huggingface/CodeT5-base"
]

# Sample text
text = """The quick brown fox jumps over the lazy dog. The dog didn't notice because it was asleep.
          Meanwhile, the fox ran away happily into the forest."""

# Run all models
for model in models:
    try:
        summarizer = pipeline("summarization", model=model)
        summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
        print(f"\nModel: {model}\nSummary: {summary[0]['summary_text']}\n")
    except Exception as e:
        print(f"\nModel: {model} - Error: {str(e)}\n")
```

```python

from transformers import pipeline
from datasets import load_dataset
from rouge_score import rouge_scorer
import nltk
from bert_score import score
import sacrebleu

nltk.download("punkt")

# Load the summarization model
summarizer = pipeline("summarization", model="facebook/bart-base")

# Load a dataset (CNN/DailyMail)
dataset = load_dataset("cnn_dailymail", "3.0.0", split="test[:100]")  # Take 100 test samples

# Store references and generated summaries
references = []
generated_summaries = []

for data in dataset:
    text = data["article"]  # Input article
    reference_summary = data["highlights"]  # Ground truth summary

    # Generate model's summary
    summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
    generated_summary = summary[0]["summary_text"]

    references.append(reference_summary)
    generated_summaries.append(generated_summary)

# Compute ROUGE Score
scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)
rouge_scores = [scorer.score(ref, gen) for ref, gen in zip(references, generated_summaries)]

avg_rouge1 = sum([score["rouge1"].fmeasure for score in rouge_scores]) / len(rouge_scores)
avg_rouge2 = sum([score["rouge2"].fmeasure for score in rouge_scores]) / len(rouge_scores)
avg_rougeL = sum([score["rougeL"].fmeasure for score in rouge_scores]) / len(rouge_scores)

print(f"ROUGE-1: {avg_rouge1:.4f}")
print(f"ROUGE-2: {avg_rouge2:.4f}")
print(f"ROUGE-L: {avg_rougeL:.4f}")

# Compute BLEU Score
bleu_score = sacrebleu.corpus_bleu(generated_summaries, [references])
print(f"BLEU Score: {bleu_score.score:.2f}")

# Compute BERTScore
P, R, F1 = score(generated_summaries, references, lang="en", model_type="bert-base-uncased")
print(f"BERTScore F1: {F1.mean().item():.4f}")
```
